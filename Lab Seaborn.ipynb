# Distribution plots
Python Seaborn module contains various functions to plot the data and depict the data variations. The seaborn.distplot() function is used to plot the distplot

**Distribution plots is for univarite plots**

```python
#seaborn.distplot()

sns.distplot(Crash_df['alcohol'],bins=7, color='purple')
plt.show()
```

# Joint plot

```python
# joint plots compares two distribtions
sns.jointplot(x='speeding', y='alcohol',data= Crash_df, kind='reg')# `kind` must be one of ['scatter', 'hist', 'hex', 'kde', 'reg', 'resid
```

```python
sns.kdeplot(Crash_df['alcohol'])
plt.show()
```
